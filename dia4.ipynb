```python
totales_categoria = total_por(gastos, "categoria")
print("Total por categoría:", totales_categoria)

totales_fecha = total_por(gastos, "fecha")
print("Total por fecha:", totales_fecha)
```

```python
montos = [g["monto"] for g in gastos]
minimo, maximo, promedio, mediana = resumen(montos)

print(f"Mínimo: {minimo:.2f}")
print(f"Máximo: {maximo:.2f}")
print(f"Promedio: {promedio:.2f}")
print(f"Mediana: {mediana:.2f}")
```

```python
pruebas = ["12,50", "$ 12.50", "", "abc", "100"]
for texto in pruebas:
    print(f"{texto!r} -> {a_float(texto)}")
```

```python
valores_ok, textos_fallidos = limpiar_montos(["12.5", "12,5", "$ 8", "", "abc", "100"])
print("Valores convertidos:", valores_ok)
print("Textos que fallaron:", textos_fallidos)
```

```python
# --- clasificar_gasto ---
assert clasificar_gasto(4.99) == "menor"
assert clasificar_gasto(5) == "medio"
assert clasificar_gasto(50) == "medio"
assert clasificar_gasto(50.01) == "mayor"
assert clasificar_gasto(120, limite_menor=10, limite_mayor=100) == "mayor"

# --- total_por ---
lista_prueba = [
    {"categoria": "a", "monto": 10},
    {"categoria": "b", "monto": 5},
    {"categoria": "a", "monto": 3},
]
assert total_por(lista_prueba, "categoria") == {"a": 13, "b": 5}

# --- resumen ---
assert resumen([1, 2, 3, 4, 5]) == (1, 5, 3.0, 3)      # cantidad impar
assert resumen([1, 2, 3, 4]) == (1, 4, 2.5, 2.5)        # cantidad par

# --- a_float ---
assert a_float("12,50") == 12.5
assert a_float("$ 12.50") == 12.5
assert a_float("") is None
assert a_float("abc") is None
assert a_float("100") == 100.0

# --- limpiar_montos ---
ok, fallidos = limpiar_montos(["12.5", "12,5", "$ 8", "", "abc", "100"])
assert ok == [12.5, 12.5, 8.0, 100.0]
assert fallidos == ["", "abc"]

print("Todas las pruebas pasaron correctamente.")
```
