```python
import numpy as np
import pandas as pd

# lista de gastos que usamos para la parte A
gastos = [
    {"fecha": "2026-09-08", "categoria": "alimentación", "descripcion": "Supermercado", "monto": 45.50},
    {"fecha": "2026-09-08", "categoria": "transporte", "descripcion": "Gasolina", "monto": 25.00},
    {"fecha": "2026-09-09", "categoria": "ocio", "descripcion": "Cine", "monto": 12.00},
    {"fecha": "2026-09-09", "categoria": "alimentación", "descripcion": "Almuerzo restaurante", "monto": 18.75},
    {"fecha": "2026-09-10", "categoria": "salud", "descripcion": "Farmacia", "monto": 30.20},
    {"fecha": "2026-09-11", "categoria": "vivienda", "descripcion": "Pago de luz", "monto": 60.00},
    {"fecha": "2026-09-11", "categoria": "transporte", "descripcion": "Taxi", "monto": 8.50},
    {"fecha": "2026-09-12", "categoria": "otros", "descripcion": "Regalo cumpleaños", "monto": 22.00},
    {"fecha": "2026-09-13", "categoria": "alimentación", "descripcion": "Café y snacks", "monto": 9.30},
    {"fecha": "2026-09-14", "categoria": "ocio", "descripcion": "Suscripción streaming", "monto": 11.99},
]

# 1. saco solo los montos de cada gasto y los meto en un arreglo de numpy
# (uso list comprehension para no tener que hacer un for aparte)
montos = np.array([gasto["monto"] for gasto in gastos])

# calculo los estadisticos que pide el ejercicio
media = np.mean(montos)
mediana = np.median(montos)
desviacion = np.std(montos)
p25 = np.percentile(montos, 25)
p50 = np.percentile(montos, 50)
p75 = np.percentile(montos, 75)

print("--- Estadisticos de mis gastos ---")
print(f"Media: {media:.2f}")
print(f"Mediana: {mediana:.2f}")
print(f"Desviación estándar: {desviacion:.2f}")
print(f"Percentil 25: {p25:.2f}")
print(f"Percentil 50: {p50:.2f}")
print(f"Percentil 75: {p75:.2f}")

# 2. con indexacion booleana filtro los gastos que estan por encima de la media
# esto me devuelve solo los valores que cumplen la condicion, sin necesidad de un for
gastos_sobre_media = montos[montos > media]

# calculo que porcentaje del total representan esos gastos
porcentaje_del_total = (gastos_sobre_media.sum() / montos.sum()) * 100

print("\n--- Gastos que superan la media ---")
print(f"Cantidad: {len(gastos_sobre_media)}")
print(f"Porcentaje del total: {porcentaje_del_total:.2f}%")

# 3. genero 1000 montos simulados usando la misma media y desviacion de mis datos
# para ver si se parece a la distribucion real
rng = np.random.default_rng(42)
montos_simulados = rng.normal(media, desviacion, 1000)
media_simulada = np.mean(montos_simulados)

print("\n--- Comparacion con datos simulados ---")
print(f"Media real: {media:.2f}")
print(f"Media simulada (1000 datos): {media_simulada:.2f}")
print(f"Diferencia: {abs(media - media_simulada):.2f}")
# la diferencia es chica porque use la misma media y desviacion para generar los datos simulados


# cargo el dataset directo desde github en formato raw
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

# 1. exploro la estructura basica del dataset antes de trabajar con el
print("\n--- Exploracion del dataset ---")
print("Dimensiones (filas, columnas):", df.shape)
print(df.head())        # primeras 5 filas para ver como vienen los datos
print(df.dtypes)         # tipo de dato de cada columna
print(df.describe())     # estadisticos generales de las columnas numericas

# 2. respondo las preguntas del ejercicio con codigo

# cuenta promedio de todo el dataset
cuenta_promedio = df["total_bill"].mean()
print(f"\nCuenta promedio: ${cuenta_promedio:.2f}")

# cuantas filas hay por dia (cuenta cuantas veces se repite cada valor de "day")
print("\nFilas por día:")
print(df["day"].value_counts())

# propina maxima y en que fila esta
propina_maxima = df["tip"].max()
fila_propina_maxima = df[df["tip"] == propina_maxima]  # filtro la fila que tiene ese valor
print(f"\nPropina máxima: ${propina_maxima:.2f}")
print(fila_propina_maxima)

# 3. creo una columna nueva calculando el porcentaje de propina sobre la cuenta
df["porcentaje_propina"] = df["tip"] / df["total_bill"] * 100
promedio_porcentaje = df["porcentaje_propina"].mean()
print(f"\nPorcentaje de propina promedio: {promedio_porcentaje:.2f}%")

# 4. filtro las mesas de 4 personas o mas y comparo su cuenta promedio con la general
mesas_grandes = df[df["size"] >= 4]
cuenta_promedio_grandes = mesas_grandes["total_bill"].mean()

print(f"\nCuenta promedio mesas grandes (4+): ${cuenta_promedio_grandes:.2f}")
print(f"Cuenta promedio general: ${cuenta_promedio:.2f}")
# es mayor la de mesas grandes, tiene logica porque piden mas entre mas personas hay

# 5. agrupo por dia y saco el promedio de la cuenta en cada uno (esto se ve mas el sabado 2)
promedio_por_dia = df.groupby("day")["total_bill"].mean()
print("\nCuenta promedio por dia:")
print(promedio_por_dia)
```

## Qué es cada columna de tips.csv
- **total_bill**: cuánto costó la cuenta en total
- **tip**: cuánto dejó de propina el cliente
- **sex**: género de quien pagó
- **smoker**: si la mesa era de fumadores o no
- **day**: qué día de la semana fue
- **time**: si fue almuerzo o cena
- **size**: cuántas personas comieron en esa mesa
