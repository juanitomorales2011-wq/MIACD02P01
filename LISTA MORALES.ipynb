```python
gastos = [
    {"fecha": "2026-09-08", "categoria": "alimentación", "descripcion": "Supermercado", "monto": 45.50},
    {"fecha": "2026-09-08", "categoria": "transporte", "descripcion": "Gasolina", "monto": 25.00},
    {"fecha": "2026-09-09", "categoria": "ocio", "descripcion": "Cine", "monto": 12.00},
    {"fecha": "2026-09-09", "categoria": "alimentación", "descripcion": "Almuerzo restaurante", "monto": 18.75},
    {"fecha": "2026-09-10", "categoria": "salud", "descripcion": "Farmacia", "monto": 30.20},
    {"fecha": "2026-09-11", "categoria": "vivienda", "descripcion": "Pago de luz", "monto": 60.00},
    {"fecha": "2026-09-11", "categoria": "transporte", "descripcion": "Taxi", "monto": 8.50},
    {"fecha": "2026-09-12", "categoria": "otros", "descripcion": "Regalo cumpleaños", "monto": 22.00},
    {"fecha": "2026-09-13", "categoria": "alimentación", "descripcion": "Café y snacks", "monto": 9.30},
    {"fecha": "2026-09-14", "categoria": "ocio", "descripcion": "Suscripción streaming", "monto": 11.99},
]

print(f"Cantidad de gastos registrados: {len(gastos)}")

# 1. Gasto total, promedio y gasto máximo

# Extraemos todos los montos en una lista, usando una comprensión de lista
montos = [gasto["monto"] for gasto in gastos]

gasto_total = sum(montos)
gasto_promedio = gasto_total / len(montos)

# Para encontrar el gasto máximo (con su descripción), usamos max() sobre
# los diccionarios completos, indicando que compare por la clave "monto"
gasto_maximo = max(gastos, key=lambda g: g["monto"])

print(f"Gasto total: ${gasto_total:.2f}")
print(f"Gasto promedio: ${gasto_promedio:.2f}")
print(f"Gasto máximo: ${gasto_maximo['monto']:.2f} -> {gasto_maximo['descripcion']}")

# 2. total_por_categoria (categoría -> suma de montos)

total_por_categoria = {}

for gasto in gastos:
    categoria = gasto["categoria"]
    monto = gasto["monto"]
    # .get(clave, 0) devuelve el valor actual de esa categoría, o 0 si aún
    # no existe en el diccionario. Así evitamos un KeyError la primera vez.
    total_por_categoria[categoria] = total_por_categoria.get(categoria, 0) + monto

print("\nTotal gastado por categoría:")
for categoria, total in total_por_categoria.items():
    print(f"  {categoria}: ${total:.2f}")


# 3. gastos_por_dia (fecha -> cantidad de gastos)

gastos_por_dia = {}

for gasto in gastos:
    fecha = gasto["fecha"]
    gastos_por_dia[fecha] = gastos_por_dia.get(fecha, 0) + 1

print("\nCantidad de gastos por día:")
for fecha, cantidad in gastos_por_dia.items():
    print(f"  {fecha}: {cantidad} gasto(s)")

# 4. Descripciones de gastos mayores a 20 (comprensión de lista)

descripciones_mayores_20 = [gasto["descripcion"] for gasto in gastos if gasto["monto"] > 20]

print("\nGastos mayores a $20:")
print(descripciones_mayores_20)

# 5. Ordenar gastos de mayor a menor monto, mostrar los 3 primeros

gastos_ordenados = sorted(gastos, key=lambda g: g["monto"], reverse=True)

print("\nTop 3 gastos más altos:")
for gasto in gastos_ordenados[:3]:
    print(f"  {gasto['descripcion']}: ${gasto['monto']:.2f} ({gasto['categoria']})")

# 6. Agregar un gasto nuevo y eliminar uno por descripción

nuevo_gasto = {
    "fecha": "2026-09-14",
    "categoria": "transporte",
    "descripcion": "Bus interprovincial",
    "monto": 15.00
}
gastos.append(nuevo_gasto)

print(f"\nGastos después de agregar uno nuevo: {len(gastos)}")

# Eliminamos un gasto por su descripción.
# Buscamos el diccionario cuya descripción coincida y lo removemos de la lista.
descripcion_a_eliminar = "Café y snacks"

gasto_a_eliminar = next(
    (gasto for gasto in gastos if gasto["descripcion"] == descripcion_a_eliminar),
    None
)

if gasto_a_eliminar is not None:
    gastos.remove(gasto_a_eliminar)

print(f"Gastos después de eliminar '{descripcion_a_eliminar}': {len(gastos)}")

# RETO: ¿en qué categoría se gastó más?

categoria_mayor_gasto = max(total_por_categoria, key=total_por_categoria.get)

print(f"\nLa categoría con mayor gasto fue: {categoria_mayor_gasto} "
      f"(${total_por_categoria[categoria_mayor_gasto]:.2f})")
```
